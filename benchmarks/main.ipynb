To set up a Julia project after cloning it, follow these steps in your terminal:

```bash
# 1. Navigate into the project folder
cd benchmarks
# 2. Open Julia
julia
# 3. Enter Package mode (press ']')
activate .
instantiate
```

Alternatively, you can do this in one command directly from your shell:

```bash
julia --project=. -e 'using Pkg; Pkg.instantiate()'
```

````julia
using Pkg

# 1. Provide the absolute path to the python executable INSIDE your venv
# For Windows: "C:\\path\\to\\venv\\Scripts\\python.exe"
# For Mac/Linux: "/path/to/venv/bin/python"
ENV["PYTHON"] = "/your/full/path/to/venv/bin/python"

# 2. Rebuild PyCall to point to this new path
Pkg.build("PyCall")
````

# Verification

````julia
using PyCall
println(PyCall.python) # Should show your venv path
println(PyCall.libpython) # Should point to the library in your venv
````
